# 🌸 Classifying the Iris Dataset with Support Vector Machines

In this notebook, we'll use Support Vector Machines (SVM) to classify the famous Iris dataset. The Iris dataset contains measurements of iris flowers and aims to classify them into one of three species:

- **Iris setosa**
- **Iris versicolor**
- **Iris virginica**

Each flower is represented by 4 features:
- Sepal length
- Sepal width
- Petal length
- Petal width

This is a multiclass classification problem.

```
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import confusion_matrix, accuracy_score
```

## Step 1: Load the Iris Dataset
We load the dataset using `load_iris()` from `sklearn.datasets`. It contains 150 samples with 4 features each and labels corresponding to 3 different species of iris flowers.

```
iris = load_iris()
X = iris.data     # Features: sepal length, sepal width, petal length, petal width
y = iris.target   # Labels: 0 = setosa, 1 = versicolor, 2 = virginica
```

## Step 2: Split the Dataset
We split the dataset into 70% training and 30% testing using `train_test_split`.

```
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

## Step 3: Train the SVM Classifier
We train an SVM classifier using the default Radial Basis Function (RBF) kernel. You can change `kernel`, `C`, and `gamma` for tuning performance.

```
clf = SVC(random_state=42)
clf.fit(X_train, y_train)
```

## Step 4: Make Predictions
We use the trained model to predict the species of the test set flowers.

```
y_pred = clf.predict(X_test)
```

## Step 5: Evaluate the Classifier
We evaluate the classifier using accuracy score and confusion matrix. These metrics tell us how well the model performed.

```
cm = confusion_matrix(y_test, y_pred)
acc = accuracy_score(y_test, y_pred)

print("Confusion Matrix:")
print(cm)
print(f"Accuracy: {acc*100:.2f}%")
```

## ✅ Summary and Next Steps
- We used an SVM classifier to predict iris species with high accuracy.
- The **confusion matrix** shows how well each class was predicted.
